# Smith-Waterman Algorithm Implementation Activity
## Introduction

Pairwise sequence alignment is a fundamental technique in bioinformatics used to compare two biological sequences, such as DNA, RNA, or proteins. It helps identify regions of similarity that may indicate functional, structural, or evolutionary relationships between the sequences.

The brute force method for sequence alignment, which involves comparing every possible alignment, is highly inefficient. For two sequences of length m and n, the time complexity would be \(T(m,n) = O(mn \cdot 2^{m+n})\), making it impractical for longer sequences.

Dynamic programming offers a more efficient solution. It breaks down the problem into smaller subproblems and stores their solutions to avoid redundant computations. This approach reduces the time complexity to \(O(mn)\).

The Smith-Waterman algorithm is a dynamic programming approach for local sequence alignment. It identifies the optimal local alignment between two sequences by comparing all possible pairs of segments from the sequences and finding the best-scoring alignment.

## Activity Overview

In this activity, you will implement the Smith-Waterman algorithm for local sequence alignment. The algorithm consists of two main parts:

Scoring matrix calculation
Traceback for optimal alignment reconstruction

The pseudocode for the Smith-Waterman algorithm is as follows:

$$
\begin{aligned}
& \text{Initialize scoring matrix } H \text{ with zeros} \\
& \text{For } i \text{ from } 1 \text{ to } m: \\
&     \text{For } j \text{ from } 1 \text{ to } n: \\
&         H[i][j] = \max \begin{cases}
&             0 \\
&             H[i-1][j-1] + \text{match/mismatch score} \\
&             H[i-1][j] + \text{gap penalty} \\
&             H[i][j-1] + \text{gap penalty}
&         \end{cases} \\
& \text{Find the highest score in } H \text{ and its position} \\
& \text{Perform traceback from the highest score position} \\
& \text{Return the optimal local alignment}
\end{aligned}
$$

## Implement Smith-Waterman algorithm


```
SmithWaterman(seq1, seq2, match, mismatch, gap)
    Initialize [len(seq1)+1] x [len(seq2)+1] numpy array as scoring matrix with first column and row equal to 0
    
    Fill scoring matrix (score_matrix) and Traceback matrix, record the position with max score (max_pos):
    for i in each row number:
        for j in each column number:
            S[i][j] = max( S[i-1][j-1] + compute_diag_score, S[i-1][j] + gap_score, H[i][j-1] + gap_score, 0 )
            T[i][j] = direction of max( S[i-1][j-1] + compute_diag_score, S[i-1][j] + gap_score, S[i][j-1] + gap_score, 0 )
    
    Traceback. Find the optimal path through scoring matrix starting at max_pos
    
```

````python
'''Helper functions'''

def matrix_init(seq1, seq2):
    ''' Initilize matrix with zeros in first column and row

    Args:
        seq1 (str): sequence 1
        seq2 (str): sequence 2
    
    Returns:
        matrix with zeros in first column and row, empty in all other cells for scoring
    '''
    rows, cols = len(seq1) + 1, len(seq2) + 1

    # fill whole matrix with None
    matrix = np.full((rows, cols), None)

    # change first row and col to 0
    matrix[0, :] = 0
    matrix[:, 0] = 0
    return matrix

def get_node_above(matrix, row, col):
    ''' Get the node directly above current node in matrix

    Args:
        matrix: scoring matrix
        row (int): row of current node
        col (int): column of current node
    
    Returns:
        value of node above current node (int)
    '''
    if row > 0:
        return matrix[row - 1][col]
    else:
        return 0  # first row and col == 0
    
def get_node_left(matrix, row, col):
    ''' Get the node directly left of current node in matrix

    Args:
        matrix: scoring matrix
        row (int): row of current node
        col (int): column of current node
    
    Returns:
        value of node left of current node (int)
    '''
    if col > 0:
        return matrix[row][col - 1]
    else:
        return 0  # first row and col == 0

def get_node_diag(matrix, row, col):
    ''' Get the node diagonal (left and up) current node in matrix

    Args:
        matrix: scoring matrix
        row (int): row of current node
        col (int): column of current node
    
    Returns:
        value of node diagonal (left and up) current node (int)
    '''
    if row > 0 and col > 0:
        return matrix[row -1][col - 1]
    else:
        return 0  # first row and col == 0
````

````python
def cal_score(matrix, seq1, seq2, i, j, match, mismatch, gap):
    '''Calculate score for position (i,j) in scoring matrix, also record move to trace back
    
    Args:
        matrix (numpy array): scoring matrix
        seq1 (str): sequence 1
        seq2 (str): sequence 2
        i (int): current row number
        j (int): current column number
        match (int): score for match
        mismatch (int): score for mismatch
        gap (int): score for gap
        
    Returns:
        score in position (i,j)    
        move to trace back: 0-END, 1-DIAG, 2-UP, 3-LEFT
        
    Pseudocode:
        Calculate scores based on upper-left, up, and left neighbors:
            diag_score = upper-left + (match or mismatch)
            up_score = up + gap
            left_score = left + gap
        score = max(0, diag_score, up_score, left_score)
        traceback = maximum direction or end
        
    '''
    # define up, left, and diagonal prev nodes
    left = get_node_left(matrix, i, j)
    up = get_node_above(matrix, i, j)
    diag = get_node_diag(matrix, i, j)

    # Calculate scores:
    # diag_score must check for matches
    if seq1[i-1] == seq2[j-1]:
        diag_score = diag + match
    else:
        diag_score = diag + mismatch

    # up and left mean there is a gap
    up_score = up + gap
    left_score = left + gap

    # get max of scores in pos i,j
    score = max(0, diag_score, up_score, left_score)
    matrix[i,j] = score
    
    # traceback
    if score == diag_score: trace = 1   # diag = 1
    elif score == up_score: trace = 2   # up = 2
    elif score == left_score: trace = 3 # left = 3
    else: trace = 0                    # end = 0

    return score, trace
````

````python
def traceback(seq1, seq2, traceback_matrix, maximum_position):
    '''Find the optimal path through scoring marix
        
        diagonal: match/mismatch
        up: gap in seq1
        left: gap in seq2
        
    Args:
        seq1 (str) : First sequence being aligned
        seq2 (str) : Second sequence being aligned
        traceback_matrix (numpy array): traceback matrix
        maximum_position (tuple): starting position to trace back from
        
    Returns:
        aligned_seq1 (str): e.g. GTTGAC
        aligned_seq2 (str): e.g. GTT-AC
        
    Pseudocode:
        while current_move != END:
            current_move = traceback_matrix[current_row][current_col]
            if current_move == DIAG:
                ...
            elif current_move == UP:
                ...
            elif current_move == LEFT:
                ...
            
    '''
    #setting up the variables to store the aligned sequences and the current position in the traceback matrix
    aligned_seq1 = ""
    aligned_seq2 = ""
    i, j = maximum_position
    #loop to trace back through the traceback matrix until reaching the end of the alignment
    while i > 0 and j > 0:
        current_move = traceback_matrix[i][j]
        #if the current move is DIAG, add the corresponding characters from seq1 and seq2 to 
        # aligned_seq1 and aligned_seq2, respectively, then move diagonally up-left in the traceback matrix
        if current_move == 1:  # DIAG
            aligned_seq1 += seq1[i - 1]
            aligned_seq2 += seq2[j - 1]
            i -= 1
            j -= 1
        #if the current move is UP, add a gap to aligned_seq2 and the corresponding character from 
        # seq1 to aligned_seq1, then move up in the traceback matrix
        elif current_move == 2:  # UP
            aligned_seq1 += seq1[i - 1]
            aligned_seq2 += '-'
            i -= 1
        #if the current move is LEFT, add a gap to aligned_seq1 and the corresponding 
        # character from seq2 to aligned_seq2, then move left in the traceback matrix
        elif current_move == 3:  # LEFT
            aligned_seq1 += '-'
            aligned_seq2 += seq2[j - 1]
            j -= 1
        else:  # END
            break
    return aligned_seq1, aligned_seq2
            
````

````python
def smith_waterman(seq1, seq2):
    '''Smith-Waterman algorithm for local alignment
    
    Args:
        seq1 (str): input seq 1
        seq2 (str): input seq 2
        match: default = +1
        mismatch: default = -1
        gap: default = -1
    
    Returns:
        aligned_seq1 (str)
        aligned_seq2 (str)
        score_matrix (numpy array): scoring matrix
    
    Pseudocode: 
        Initialize [len(seq1)+1] x [len(seq2)+1] numpy array as scoring matrix with first column and row equal to 0
        Fill scoring matrix (score_matrix) and Traceback matrix, record the position with max score (max_pos):
        for i in each row number:
            for j in each column number:
                S[i][j] = max( S[i-1][j-1] + compute_diag_score, S[i-1][j] + gap_score, H[i][j-1] + gap_score, 0 )
                T[i][j] = direction of max( S[i-1][j-1] + compute_diag_score, S[i-1][j] + gap_score, S[i][j-1] + gap_score, 0 )
        Traceback. Find the optimal path through scoring matrix starting at max_pos
    '''
    # initialize traceback and score matrix
    traceback_matrix = matrix_init(seq1, seq2)
    score_matrix = matrix_init(seq1, seq2)

    # define scoring rules
    match = 2
    mismatch = 1
    gap = 0

    # variable for max score
    max_score = 0
    
    for i in range(1, len(seq1) + 1):
        for j in range(1, len(seq2) + 1):
            # get max score (up/left/diag) and trace and populate current node
            score, trace = cal_score(score_matrix, seq1, seq2, i, j, match, mismatch, gap)
            score_matrix[i][j] = score
            traceback_matrix[i][j] = trace
            # check for max score
            if score_matrix[i][j] > max_score: max_score = score_matrix[i][j]
    
    # trace back through the matrix for the optimal path starting at max_pos (maybe multiple)
    # np.where will return position of the max, even if there are multiple
    # max_row and max_col will have x,y vals for the position in list format
    max_row, max_col = np.where(score_matrix == max_score)

    # loop through length of max_row/max_col to find max_pos (maybe multiple)
    tracebacks = []
    for i in range(len(max_row)):
        max_pos = (max_row[i], max_col[i])
        # traceback starting at max_pos
        tracebacks.append(traceback(seq1, seq2, traceback_matrix, max_pos))

    # get aligned seqs
    aligned_seq1, aligned_seq2 = tracebacks[0]

    return aligned_seq1, aligned_seq2,score_matrix
````
